# Guia do projeto

Este conjunto de cadernos reúne a análise de dados públicos do TCC "Investigação de assinaturas de vesículas extracelulares associadas à modulação fenotípica de células mamárias no contexto do câncer de mama". Ele atende ao quarto objetivo específico do projeto: investigar, com dados públicos, os desafios de acesso ao diagnóstico e ao acompanhamento do câncer de mama no sistema público de saúde brasileiro, comparando as cinco regiões do país.

Cada caderno responde a uma pergunta. Todos seguem a mesma estrutura: primeiro dizem qual é a pergunta e de onde vêm os dados, depois mostram o cálculo passo a passo e, por fim, explicam como ler o resultado e quais cuidados tomar ao citá-lo no texto do TCC.

## Ordem de leitura

| Caderno | Pergunta |
|---|---|
| 01 Fontes e qualidade dos dados | De onde vêm os dados e o quanto podemos confiar em cada um? |
| 02 Perfil das pacientes | Quem são as mulheres com câncer de mama atendidas no país? |
| 03 Rastreamento e diagnóstico | Quantas mamografias são feitas, para quem, com que resultado e com que demora? |
| 04 Estádio e tipo histológico | Em que fase da doença o câncer é descoberto, e isso muda com região, idade, raça/cor e escolaridade? |
| 05 Subtipos moleculares | Qual a proporção de tumores luminais, HER2 positivos e triplo-negativos, e como ela varia? |
| 06 Tempo até o tratamento | Quanto tempo passa entre a consulta, o diagnóstico e o início de cada tratamento? |
| 07 Sazonalidade dos diagnósticos | Há meses do ano com mais diagnósticos? |
| 08 Rede pública, filantrópica e privada | Em que tipo de hospital as mulheres são tratadas, e isso se associa a diferenças de estádio e de tempo? |
| 09 Equipamentos e serviços | Quantos mamógrafos, tomógrafos e hospitais oncológicos existem, quantos atendem o SUS e onde estão? |
| 10 Deslocamento para o tratamento | Quanto as mulheres viajam para cada etapa do cuidado, da mamografia à radioterapia? |
| 11 Mortalidade | Como a mortalidade evolui por região e por idade? |
| 12 Síntese para o TCC | Quais resultados entram no texto, com que números e com que ressalvas? |

Os cadernos 02 a 11 podem ser lidos em qualquer ordem depois do 01. O 12 resume todos.

## Como o projeto está organizado

| Pasta ou arquivo | Conteúdo |
|---|---|
| `notebooks/` | Os cadernos, que são o produto principal da análise |
| `src/mama/` | O código reutilizado pelos cadernos, organizado como um pacote Python chamado `mama` |
| `src/mama/download/` | Funções que baixam os arquivos do INCA e do DATASUS |
| `src/mama/fontes/` | Um módulo por base de dados; cada um lê os arquivos brutos, filtra câncer de mama em mulheres e padroniza as colunas |
| `src/mama/analises/` | Um módulo por tema (perfil, rastreamento, estadiamento, subtipos, tempos, sazonalidade, rede, equipamentos, deslocamento, mortalidade) |
| `src/mama/codigos.py` | Tabelas de códigos dos sistemas oficiais (UF, raça/cor, escolaridade, habilitações, procedimentos) |
| `src/mama/graficos.py` | Estilo único de todas as figuras |
| `scripts/` | Três comandos: um baixa os dados, outro constrói as bases limpas e o terceiro executa todos os cadernos |
| `data/raw/` | Arquivos exatamente como foram baixados (não vão para o GitHub, são grandes) |
| `data/interim/` | Bases limpas em formato parquet, uma por fonte |
| `data/processed/` | Tabelas finais em CSV, prontas para abrir no Excel |
| `figures/` | Figuras em PNG com 300 dpi, numeradas pelo caderno que as gera |
| `tests/` | Testes automáticos das funções de padronização |

A separação tem um motivo prático. Se uma regra de limpeza mudar, ela muda em um único lugar (`src/mama`) e todos os cadernos passam a usá-la. Os cadernos ficam curtos e concentrados na explicação.

## Como reproduzir tudo do zero

Os comandos abaixo devem ser rodados no terminal, dentro da pasta `dados-publicos`. O primeiro cria o ambiente Python e instala o pacote. O segundo baixa cerca de 6 GB de arquivos públicos e pode levar algumas horas. O terceiro constrói as bases limpas, o que leva em torno de 40 minutos. O quarto executa todos os cadernos em ordem, regravando tabelas e figuras.

```bash
uv venv .venv --python 3.12
uv pip install --python .venv/bin/python -e ".[dev]"
.venv/bin/python scripts/baixar_dados.py
.venv/bin/python scripts/construir_bases.py
.venv/bin/python scripts/executar_cadernos.py
.venv/bin/python -m pytest -q tests
```

Os downloads podem ser interrompidos e retomados: arquivos que já existem são pulados.

## Vocabulário

| Termo | Significado neste trabalho |
|---|---|
| Estádio | Fase da doença no momento do diagnóstico, de 0 a IV, definida pelo sistema TNM (tamanho do tumor, linfonodos e metástase). É o termo usado pelo INCA e pela tradução oficial da classificação TNM; "estágio" é a forma coloquial. Estadiamento é o processo de classificar o estádio. |
| Estádio avançado | Estádios III e IV. É o indicador usado na literatura brasileira para diagnóstico tardio. |
| Caso analítico | Paciente que teve diagnóstico e/ou tratamento planejado e feito no hospital que registrou o caso. São os casos usados para avaliar a assistência. |
| RHC | Registro Hospitalar de Câncer, mantido por cada hospital e consolidado pelo INCA. |
| Painel de Oncologia | Base do DATASUS que junta os registros do SUS para monitorar a Lei 12.732/2012, que garante o início do tratamento em até 60 dias após o diagnóstico. |
| APAC | Autorização de Procedimento de Alta Complexidade, documento que o SUS usa para pagar quimioterapia e radioterapia. |
| CNES | Cadastro Nacional de Estabelecimentos de Saúde: lista de todos os serviços de saúde, seus equipamentos e habilitações. |
| UNACON e CACON | Hospitais habilitados pelo Ministério da Saúde para tratar câncer (Unidade e Centro de Assistência de Alta Complexidade em Oncologia). |
| Atenção primária | Postos de saúde, Unidades Básicas de Saúde (UBS) e Unidades de Saúde da Família, a porta de entrada do SUS. |
| SIM | Sistema de Informações sobre Mortalidade, com todas as declarações de óbito do país. |
| SISCAN | Sistema de Informação do Câncer, onde o SUS registra mamografias e laudos de biópsia de mama. |
| BI-RADS | Escala de 0 a 6 usada no laudo da mamografia; as categorias 4 e 5 indicam achado suspeito de câncer e levam à biópsia. |
| Imuno-histoquímica | Exame feito na biópsia que mede os receptores hormonais e o HER2, definindo o subtipo molecular do tumor. |
| Taxa padronizada por idade | Taxa calculada como se todas as regiões tivessem a mesma estrutura etária. Permite comparar regiões com populações mais jovens ou mais velhas. |
